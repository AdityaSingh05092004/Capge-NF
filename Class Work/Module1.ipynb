**WAP of grade system of 5 subjects by taking user input**

```python
a=input("Enter your name ")
print("Enter marks of all subjects ")
b=int(input("Enter marks of English:"))
c=int(input("Enter marks of Maths:"))
d=int(input("Enter marks of Science:"))
e=int(input("Enter marks of Hindi:"))
f=int(input("Enter marks of Computer:"))
totalmarks =b+c+d+e+f
print("Total marks:",totalmarks)
per=totalmarks/500*100
print("Percentage of all 5 subjects ",per,"%")
if per>=90 and per<=100:
    print("grade= A")
elif per>=80 and per<90:
    print("grade= B")
elif per>=70 and per<80:
    print("grade= C")
elif per>=60 and per<70:
    print("grade= D")
else:
    print("grade= F")
```

```python
def Sum(n):
  Tsum=0
  for i in range(0,n+1):
    Tsum=Tsum+i
  return Tsum
n=int(input("Enter a number:"))
Sum(n)
```

```python
def Fact(n):
  Fac=1
  for i in range(1,n+1):
    Fac=Fac*i
  return Fac
Fact(5)
```
