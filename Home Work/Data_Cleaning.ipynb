**Load Dataset**

```python
# Load the messy employment dataset into a DataFrame
df = pd.read_csv("Messy_Employment_India_Dataset.csv")
df.head()       # Display first five records
```

**Identify and Handle Missing Values**

```python
# Check missing values in each column
df.isnull().sum()
```

```python
# Fill missing values in numerical columns using median
df["Years of Experience"] = df["Years of Experience"].fillna(df["Years of Experience"].median())
df["Monthly Salary (INR)"] = df["Monthly Salary (INR)"].fillna(df["Monthly Salary (INR)"].median())
```

```python
# Fill missing values in categorical columns using mode
categorical_cols = ["Status","Age Group","Education","Industry","Location","AI Risk"]
for i in categorical_cols:
    df[i] = df[i].fillna(df[i].mode()[0])
```

```python
# Verify missing values after treatment
df.isnull().sum()
```

**Remove Duplicate Records**

```python
# Count duplicate records
print(df.duplicated().sum())
```

**Correct Inconsistent Formatting**

```python
# Remove leading and trailing spaces from text columns
text_cols = df.select_dtypes(include="object").columns
for i in text_cols:
    df[i] = df[i].str.strip()
```

```python
# Standardize status values
df["Status"] = df["Status"].str.lower()
df["Status"] = df["Status"].replace({"employed":"Employed","unemployed":"Unemployed"})
```

```python
# Standardize AI Risk values
df["AI Risk"] = df["AI Risk"].str.lower()
df["AI Risk"] = df["AI Risk"].replace({
    "low": "Low",
    "moderate": "Moderate",
    "medium": "Moderate",
    "high": "High"
})
```

```python
# Standardize age group formats
df["Age Group"] = df["Age Group"].replace({
    "25_34": "25-34",
    "18 to 24": "18-24",
    "35 - 44": "35-44"
})
```

```python
# Standardize education values
df["Education"] = df["Education"].replace({
    "high school": "High School",
    "Master": "Masters",
    "Ph.D": "PhD",
    "Bachelors": "Bachelor's"
})
```

```python
# Standardize industry names
df["Industry"] = df["Industry"].replace({"Tech": "Technology","Fintech": "Finance"})
```

**Fix Data Type**

```python
# Convert experience column to numeric type
df["Years of Experience"] = pd.to_numeric(
    df["Years of Experience"],
    errors="coerce"
)
```

```python
# Convert salary column to numeric type
df["Monthly Salary (INR)"] = pd.to_numeric(
    df["Monthly Salary (INR)"],
    errors="coerce"
)
```

```python
# Convert date column to datetime
df["Date Recorded"] = pd.to_datetime(
    df["Date Recorded"]
)
```

```python
# Verify data types
df.dtypes
```

**Detect and Handle Outliers**

```python
# Handle outliers in Years of Experience and Salary using IQR
num_cols = ["Years of Experience", "Monthly Salary (INR)"]
for col in num_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = np.where(df[col] < lower, lower, df[col])
    df[col] = np.where(df[col] > upper, upper, df[col])
```

**Rename and Standardize Column Names**

```python
# Standardize column names
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("(", "", regex=False)
      .str.replace(")", "", regex=False)
)
```

```python
# View updated column names
print(df.columns)
```
