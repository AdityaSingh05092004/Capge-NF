```python
# Q1 Create a Dictionary from Two Lists
keys=["name","age","city"]
values=["Aditya",20,"Delhi"]
a=dict(zip(keys,values))
print(a)
```

```python
# Q2 Merge Two Dictionaries
a={'name':'Aditya','age':20,'city':'Delhi'}
b={'Phone No.':123456789}
a.update(b)
print("Merged Dictionary:",a)
```

```python
# Q4 Check if One Set is Subset of Another
a={1,2,3}
b={1,2,3,4,5}
if(a.issubset(b)):
    print("a is a subset of b")
else:
    print("a is not a subset of b")
```

```python
# Q4 Check Common Elements in Two Lists Using Sets
list1=[1,2,3,4]
list2=[4,5,6,7]
a=set(list1)
b=set(list2)
c=a.intersection(b)
if(c):
    print("Common element exists",c)
else:
    print("No common elements")
```
