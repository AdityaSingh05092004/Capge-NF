```python
# Q1 To check anagrams 
def check_anagram(a,b):                  #Anagrams are words formed by rearranging the letters of another word
    a=a.replace(" ","").lower()
    b=b.replace(" ","").lower()          # replace()-Removes spaces from strings
    return sorted(a)==sorted(b)

a=input("Enter 1st String")
b=input("Enter 2nd String")
check_anagram(a,b)
```

```python
# Q4 To Check Whether a Number is an Armstrong
def is_armstrong(number):                 # An Armstrong number is a number equal to the sum of its digits raised to the power of total digits
    digits = str(number)
    power = len(digits)
    
    total = sum(int(i) ** power for i in digits)
    return total == number

# Example
print(is_armstrong(153))
```

```python
# Q5 Find common elements 
def common(list1,list2,list3):
    return set(list1).intersection(list2,list3)

list1=[1,2,3,4]
list2=[2,3,5,6]
list3=[2,3,7,8]
common(list1,list2,list3)
```
