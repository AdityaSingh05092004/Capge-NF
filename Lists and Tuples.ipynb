```python
# Q1 Largest and Smallest Elements in a List
list1 = [10, 45, 2, 78, 23]
a=max(list1)
b=min(list1)
print("Largest Element:",a)
print("Smallest Element:",b)
```

```python
# Q2 Remove Duplicate Elements from a List
list1=[1, 2, 2, 3, 4, 4, 5]
a=[]
for i in list1:
    if(i not in a):
        a.append(i)
print("List after removing duplicates:",a)
# by using set method 
b=list(set(list1))
print("List after removing duplicates:",b)
```

```python
# Q3 Reverse a List Without Using reverse()
list1 = [10, 45, 2, 78, 23]
print("Reversed List:",list1[::-1])
```

```python
# Q4 Count Even and Odd Numbers
list1 = [10, 45, 2, 78, 23]
a=0
b=0
for i in list1:
    if(i%2==0):
        a=a+1
    else:
        b=b+1
print("Even Numbers:",a)
print("Odd Numbers:",b)
```

```python
# Q5 Merge Two Lists and Sort
a=[1,2,3]
b=[4,5,6]
print("Merged List:",a+b)
a.extend(b)
print("Merged List:",a)
```

```python
# Q6 Second Largest Number
list1 = [10, 45, 2, 78, 23]
list1.sort()
print("Second Largest Number of List:",list1[-2])
```

```python
# Q7 Check Element Exists
tup=(10, 20, 30, 40)
a=int(input("Enter element to search:"))
if(a in tup):
    print("Element exists in tuple")
else:
    print("Element does not exist")
```

```python
# Q8 Count Occurence 
tup = (1, 2, 3, 2, 4, 2)
a=int(input("Enter element:"))
b=tup.count(a)
print("Occurrence of Element:",b)
```

```python
# Q9 Sort List of Tuples
a=[(10,2),(2,1),(3,1),(2,0),(4,2),(2,1)]
a.sort()
print("Sorted List of Tuples:",a)
```

```python
# Q10 Convert Tuple to List and List to Tuple
tup=(1,2,3,4)
list1=[5,6,7,8]
a=list(tup)
print("Tuple to List:",a)
b=tuple(list1)
print("List to Tuple:",b)
```
